```python
import pandas as pd
import numpy as np

data = pd.read_csv('data.csv')

# data = {
#    'AgeGroup': ['20-29', '30-39', '20-29', '40-49', '30-39', '40-49'],
#    'MonthlyIncome': [3000, 5000, 3500, 8000, 5500, 8200],
#    'Attrition': ['Yes', 'No', 'No', 'No', 'Yes', 'No'],
#    'BusinessTravel': ['Travel_Rarely', 'Travel_Frequently', 'Travel_Rarely', 'Non-Travel', 'Travel_Frequently', 'Travel_Rarely']
# }
df = pd.DataFrame(data)
summary_stats = df.groupby('AgeGroup')['MonthlyIncome'].agg(['mean', 'median', 'min', 'max', 'std'])
print(summary_stats)

df["BusinessTravel"] = df["BusinessTravel"].replace({"Travel_Rarely": 1, "Travel_Frequently": 0, "Non-Travel": 2})
df["Attrition"] = df["Attrition"].replace({"Yes": 1, "No": 0})

attrition_numeric_list = df["Attrition"].tolist()

df.head()
print(f"List of numeric values for Attrition: {attrition_numeric_list}")
```

```python
import pandas as pd

iris_data = pd.read_csv(r"F:\DSBDA\DSBDAL CODES\iris.csv", names = ['Sepal_Length', 'Sepal_Width', 'Petal_Length', 'Petal_Width', 'Species'])
iris_df = pd.DataFrame(iris_data)

species_list = ['Iris-setosa', 'Iris-versicolor', 'Iris-virginica']

for species in species_list:
    print(f"--- Statistical Details for {species} ---")
    species_data = iris_df[iris_df['Species'] == species]
    stats = species_data.describe()
    print(stats)
```
