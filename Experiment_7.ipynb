```python
import nltk
import math
import pandas as pd
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

# ---------------------------------------------------------
# SETUP: Download required NLTK data (run once)
# ---------------------------------------------------------
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('omw-1.4')

# ---------------------------------------------------------
# PART 1: Document Preprocessing
# ---------------------------------------------------------

#with open('text.txt', 'r') as file:
#    text = file.read()

# Sample Document
text = "Tokenization is the first step in text analytics. The process of breaking down a text paragraph into smaller chunks such as words or sentences is called Tokenization."
print("--- ORIGINAL TEXT ---")
print(text, "\n")

# 1. Tokenization
tokens = word_tokenize(text)
print("1. Tokenization (Words):")
print(tokens, "\n")

# 2. POS Tagging
pos_tags = nltk.pos_tag(tokens)
print("2. POS Tagging:")
print(pos_tags, "\n")

# 3. Stop Words Removal
stop_words = set(stopwords.words('english'))
# Removing punctuation as well for cleaner processing
filtered_tokens = [w for w in tokens if w.lower() not in stop_words and w.isalnum()]
print("3. Stop Words Removal:")
print(filtered_tokens, "\n")

# 4. Stemming
ps = PorterStemmer()
stemmed_words = [ps.stem(w) for w in filtered_tokens]
print("4. Stemming:")
print(stemmed_words, "\n")

# 5. Lemmatization
lemmatizer = WordNetLemmatizer()
lemmatized_words = [lemmatizer.lemmatize(w) for w in filtered_tokens]
print("5. Lemmatization:")
print(lemmatized_words, "\n")


# ---------------------------------------------------------
# PART 2: TF-IDF Calculation
# ---------------------------------------------------------
print("--- TF-IDF CALCULATION ---")
# Two sample documents for TF-IDF
docA = "Jupiter is the largest planet"
docB = "Mars is the fourth planet from the sun"

# Preprocess: lowercase and split
bowA = docA.lower().split()
bowB = docB.lower().split()

# Create a collection of all unique words
unique_words = set(bowA).union(set(bowB))

# Count word occurrences in each document
word_countA = dict.fromkeys(unique_words, 0)
for word in bowA:
    word_countA[word] += 1

word_countB = dict.fromkeys(unique_words, 0)
for word in bowB:
    word_countB[word] += 1

# Function to calculate Term Frequency (TF)
def compute_tf(word_dict, bow):
    tf_dict = {}
    bow_count = len(bow)
    for word, count in word_dict.items():
        tf_dict[word] = count / float(bow_count)
    return tf_dict

tfA = compute_tf(word_countA, bowA)
tfB = compute_tf(word_countB, bowB)

# Function to calculate Inverse Document Frequency (IDF)
def compute_idf(doc_list):
    idf_dict = {}
    N = len(doc_list)
    # Count how many documents contain each word
    idf_dict = dict.fromkeys(doc_list[0].keys(), 0)
    for doc in doc_list:
        for word, val in doc.items():
            if val > 0:
                idf_dict[word] += 1
    
    # Calculate log ratio
    for word, val in idf_dict.items():
        idf_dict[word] = math.log(N / float(val))
    return idf_dict

idfs = compute_idf([word_countA, word_countB])

# Function to calculate TF-IDF
def compute_tfidf(tf_bow, idfs):
    tfidf = {}
    for word, val in tf_bow.items():
        tfidf[word] = val * idfs[word]
    return tfidf

tfidfA = compute_tfidf(tfA, idfs)
tfidfB = compute_tfidf(tfB, idfs)

# Display the results nicely using pandas
df_tfidf = pd.DataFrame([tfidfA, tfidfB], index=['Document A', 'Document B'])
print(df_tfidf.T) # Transpose for easier reading
```
